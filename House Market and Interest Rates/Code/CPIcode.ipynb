```python
import matplotlib.pyplot as plt

plt.close("all")

periods = {}

with open("CPI Cleaned.csv") as file:
    data = file.read()
    lines = data.split("\n")
    lines.pop(0) 
    lines.pop(-1)  
    for line in lines:
        separated = line.split(",") 
        if separated[0][0:4] in periods: 
            values = periods[separated[0][0:4]]
            values.append(float(separated[1]))
        else:
            periods[separated[0][0:4]] = [float(separated[1])]

def get_mean(list):
    total = len(list)
    sum = 0
    for item in list:
        sum += item
    return sum/total

xValues = [] 
yValues = []

values = list(periods.values())

for value in values:
    yValues.append(get_mean(value))

keys = list(periods.keys())
for key in keys: 
    xValues.append(key)

plt.figure(figsize=(10, 6))
plt.plot(xValues, yValues, "y-", label="Average Annual Customer Price Index (CPI)")
plt.ylabel("Average CPI", fontsize=15.0)
plt.xlabel("Year", fontsize=15.0)
plt.legend(shadow=True, framealpha=0.7)
plt.xticks(xValues, rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
